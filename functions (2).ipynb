1) What is the difference between a function and a method in Python?
- Definition : Function is a block of reusable code that performs a specific task while Method is a function that is associated with an object .
- How it's called : Functions are directly called by function name i.e function_name() while Methods are called on an object using dot notation i.e object.method_name().
-  First argument :No implicit argument is passed in functions . The first argument is automatically the object itself.
- Functions are not tied to any object or class while Methods are tied to an object or class .


```python
# functions eg :
def sub_num(a,b):
  return a-b
result=sub_num(5,3)
print(f"{5}-{3}={result}")

# Method eg:
class cal:
  def add(self,a,b):
    return a+b
add=cal()
result=add.add(5,3)
print(f"{5}+{3}={result}")
```

6) . Explain the concept of generators in Python and how they are defined?
- A generator in Python is a special type of iterator that allows you to generate values on the fly rather than storing them all in memory at once.
- Generators are used to produce a sequence of values lazily — meaning they yield one item at a time only when requested.
- A generator function is defined just like a normal function, but instead of return, it uses the yield keyword.


```python
# eg of generators:
def even_numbers():
    n = 0
    while True:
        yield n
        n += 2

evens = even_numbers()
for _ in range(5):
    print(next(evens))
```

7) . What are the advantages of using generators over regular functions?
- Advantages of using generators over regular functions are :
- a) Memory Efficient: Don’t store all values in memory.
- b) Lazy Evaluation: Produce values only when needed.
- c) Faster for large data: Start yielding results immediately.
- d) Simpler Syntax: No need for iterator classes.
- e) Can handle infinite sequences.

```python
# eg :
def generate_numbers(n):
    i = 1
    while i <= n:
        yield i # yield i one by one
        i += 1

for num in generate_numbers(5):
    print(num)
```

10) What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?
- map(): It applies the given function to all iterables and return a new list.
- reduce():It applies some other function to a list that are passed as a parameter to it, and finally return a single value.
- filter():It is used to filter the given iterables with the help of another function passed as an argument to test all the elements to be True or False.

```python
# map eg:
addn = lambda x,y : x+y
print(addn(3,5))

# reduce eg:
from functools import reduce
nums = [1, 2, 3, 4, 5]
print(reduce(lambda x, y: x+y, nums))

# filter eg:
print(list(filter(lambda x: x%2==0, nums)))
```

4) . Write a Python function that checks if a given number is prime or not from 1 to 200?

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

for num in range(1, 201):
    if is_prime(num):
        print(num, end=" ")
```

5)  Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.?

```python
class Fibonacci:
    def __init__(self, n):
        self.n = n          # total number of terms
        self.a, self.b = 0, 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count < self.n:
            result = self.a
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return result
        else:
            raise StopIteration

for num in Fibonacci(10):
    print(num, end=" ")
```

6). Write a generator function in Python that yields the powers of 2 up to a given exponent.?

```python
def powers_of_two(n):
    for i in range(n + 1):
        yield 2 ** i

for num in powers_of_two(5):
    print(num, end=" ")
```

7) . Implement a generator function that reads a file line by line and yields each line as a string.?

```python
with open('sample.txt', 'w') as f:
    f.write("Hello world\nThis is a test file\nPython is awesome!")

def read_file_line_by_line(filename):
    with open(filename, 'r') as file:
        for line in file:
            yield line.strip()
for line in read_file_line_by_line('sample.txt'):
    print(line)

```

```python
text = "Hello World"
# Remove vowels using filter and lambda
result = ''.join(filter(lambda ch: ch.lower() not in 'aeiou', text))

print(result)
```
